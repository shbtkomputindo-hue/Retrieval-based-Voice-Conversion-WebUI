```python
# prompt: Slice audio

# Install pydub if not already installed
!pip install pydub

from pydub import AudioSegment
import math
import os # Import os to use os.path.join

# Assuming the uploaded file is in the current directory where the notebook is run
# Make sure the path is correct based on where your audio file is located after upload
audio_file_path = '/content/Vocalku.wav' # Adjust this path if the file is elsewhere

# Check if the file exists
if not os.path.exists(audio_file_path):
    print(f"Error: Audio file not found at {audio_file_path}")
else:
    # Load the audio file
    audio = AudioSegment.from_file(audio_file_path)

    # Define the slice duration in milliseconds (e.g., 10 seconds)
    slice_duration_ms = 10 * 1000

    # Calculate the number of slices
    num_slices = math.ceil(len(audio) / slice_duration_ms)

    # Create a directory for sliced audio
    output_dir = '/content/rvc_output' # Redefine output_dir if needed
    sliced_audio_dir = os.path.join(output_dir, 'sliced_audio')
    os.makedirs(sliced_audio_dir, exist_ok=True)

    print(f"\nSlicing audio into {num_slices} chunks...")

    # Slice and export the audio
    for i in range(num_slices):
        start_time = i * slice_duration_ms
        end_time = (i + 1) * slice_duration_ms
        # Ensure the end_time does not exceed the audio length
        if end_time > len(audio):
            end_time = len(audio)
        slice = audio[start_time:end_time]

        # Define the output filename
        slice_filename = os.path.join(sliced_audio_dir, f'slice_{i+1:03d}.wav')

        # Export the slice
        slice.export(slice_filename, format="wav")
        print(f"Exported: {slice_filename}")

    print("\nAudio slicing complete.")
    print(f"Sliced audio files are saved in: {sliced_audio_dir}")
```
